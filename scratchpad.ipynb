```python
P_list = [(0,0),(0,1),(1,1),(2,0)]

def bezier_curve(P_list):
    N = len(P_list) - 1
    t = np.arange(0, 1, 0.02)
    t = np.append(t,1)
    T = len(t)

    T_mat = np.zeros((N + 1,T))
    for i in range(N + 1):
        T_mat[i][:] = math.comb(N, i)*t**i * (1 - t)**(N - i)

    P_mat = np.array(P_list).T
    b_curve = P_mat@T_mat 
    return b_curve
```
